```python
import numpy as np
import pandas as pd
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("property_prices_100.csv")
```

```python
new_area = df.drop('price', axis='columns')
```

```python
new_price = df.drop('area', axis='columns')
```

```python
reg = linear_model.LinearRegression()
reg.fit(new_area, new_price)
```

```python
reg.predict([[4000]])
```

```python
%matplotlib inline
plt.xlabel('area')
plt.ylabel('price')
plt.scatter(df.area, df.price, color='red', marker='x') 
plt.show()
```

```python
df.predict = pd.read_csv('areas.csv')
```

```python
reg.predict(df.predict)
```
